```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Concatenate, Flatten
from tensorflow.keras.utils import plot_model
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

print("TensorFlow Version:", tf.__version__)
```

```python
np.random.seed(42)

n_samples = 1000
salary = np.random.normal(50000, 15000, n_samples)
height = np.random.normal(170, 10, n_samples)
marital_status = np.random.randint(0, 2, n_samples) # 0 or 1

X = np.column_stack((salary, height, marital_status))
age = np.random.normal(30, 8, n_samples)
gender = np.random.randint(0, 2, n_samples)

y_age = age
y_gender = gender

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_age_train, y_age_test, y_gender_train, y_gender_test = train_test_split(
    X_scaled, y_age, y_gender, test_size=0.2, random_state=42
)

print("X_train shape:", X_train.shape)
```

```python
input_layer = Input(shape=(3,), name='input_features')

x = Dense(64, activation='relu', name='hidden_layer_1')(input_layer)
x = Dense(64, activation='relu', name='hidden_layer_2')(x)
age_branch = Dense(32, activation='relu', name='age_hidden')(x)
output_age = Dense(1, activation='linear', name='age_output')(age_branch)
gender_branch = Dense(32, activation='relu', name='gender_hidden')(x)
output_gender = Dense(1, activation='sigmoid', name='gender_output')(gender_branch)

model = Model(
    inputs=input_layer,
    outputs=[output_age, output_gender],
    name='multi_output_functional_model'
)
model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss={
        'age_output': 'mse',
        'gender_output': 'binary_crossentropy'
    },
    loss_weights={
        'age_output': 1.0,
        'gender_output': 1.0
    },
    metrics={
        'age_output': 'mae',
        'gender_output': 'accuracy'
    }
)
```

```python
history = model.fit(
    X_train,
    [y_age_train, y_gender_train],
    epochs=20,
    batch_size=32,
    validation_split=0.2
)
```

```python

predictions = model.predict(X_test)

predicted_ages = predictions[0]
predicted_genders = predictions[1]

print("Sample Predicted Ages:", predicted_ages[:5].flatten())
print("Sample Predicted Genders (Probabilities):", predicted_genders[:5].flatten())
```
